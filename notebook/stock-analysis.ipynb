```python
import pandas as pd


stock = pd.read_csv('./data/aapl.csv')
stock
```

```python
stock['Year'] = stock['Date'].str[:4]
stock['Month'] = stock['Date'].str[5:7]
stock['Day'] = stock['Date'].str[8:10]
stock.tail()
```

```python
from matplotlib import pyplot as plt


for year in stock['Year'].unique():
    yearly_data = stock[stock['Year'] == year]
    plt.plot(yearly_data.index, yearly_data['Close'], label=f'{year}')

# Customize the plot
plt.title(f'AAPL Yearly Stock Price')
plt.xlabel('Date')
plt.ylabel('Closing Price (USD)')
plt.legend(title='Year')
plt.grid(True)
plt.show()
```
